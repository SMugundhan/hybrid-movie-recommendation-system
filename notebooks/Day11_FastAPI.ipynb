Step 7 — Recommendation route

@app.get("/recommend/{user_id}")

def recommend_movies(
    user_id:int
):

    liked_movies = df[
        (
            df['user_id']==user_id
        )
    ].sort_values(
        'rating',
        ascending=False
    )['title'].head(5).tolist()

    recommendations = hybrid_recommendation(
        user_id,
        liked_movies,
        ratings,
        movies,
        similarity,
        model
    )

    output=[]

    for _,row in recommendations.iterrows():

        movie_name = movies[
            movies['movie_id']==row['movie_id']
        ]['title'].values[0]

        output.append({

            "movie":movie_name,

            "score":float(
                row['final_score']
            )
        })

    return output
