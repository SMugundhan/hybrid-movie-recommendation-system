FROM python:3.11-slim

---

🧠 Understanding Every Line

FROM

FROM python:3.11-slim

Starts from an existing image.

Think:

Instead of building Linux + Python from scratch,
we reuse a ready-made Python image.


---

WORKDIR

WORKDIR /app

Creates:

/app

inside container.

All future commands run there.


---

COPY requirements.txt

COPY requirements.txt .

Copies:

requirements.txt

from your laptop into container.


---

RUN

RUN pip install --no-cache-dir -r requirements.txt

Installs all libraries.

Equivalent to:

pip install -r requirements.txt


---

COPY . .

COPY . .

Copies entire project.


---

EXPOSE

EXPOSE 8001

Documents that FastAPI uses port 8001.


---

CMD

CMD ["uvicorn","api.app:app","--host","0.0.0.0","--port","8001"]

Runs automatically when container starts.

🧠 Small Important Concept

Before we build:

A Docker Image is created in layers:

Layer 1 -> Python Image
Layer 2 -> Requirements
Layer 3 -> Installed Packages
Layer 4 -> Project Files

This is why Docker rebuilds quickly later.

If only your code changes:

Layer 1 -> Reuse
Layer 2 -> Reuse
Layer 3 -> Reuse
Layer 4 -> Rebuild

Very important optimization concept.
