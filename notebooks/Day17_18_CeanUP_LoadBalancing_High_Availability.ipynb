Good answer, but let's refine it into an interview-quality answer.

What you said is mostly describing a weakness of content-based filtering.

The stronger answer is:

Why not only Collaborative Filtering (SVD)?

Collaborative Filtering learns from user behavior and ratings.

However, it struggles with:
1. Cold Start Problem
   - New movies have few or no ratings.
   - New users have little interaction history.

2. Data Sparsity
   - Most users rate only a small fraction of movies.

3. Limited Content Understanding
   - It knows users liked similar items,
     but not WHY they liked them.


---

Why not only Content-Based Filtering?

Content-Based Filtering recommends movies
similar to what the user already likes.

However:

1. Over-specialization
   - User gets more of the same type of movies.

2. Limited Discovery
   - Hard to recommend something outside
     the user's historical preferences.

3. Doesn't leverage community behavior.


---

Why Hybrid?

Hybrid combines both approaches.

Collaborative Filtering:
    Learns from similar users.

Content-Based Filtering:
    Learns from movie characteristics.

Together:
    Better personalization.
    Better recommendation quality.
    Reduced weaknesses of each individual method.
