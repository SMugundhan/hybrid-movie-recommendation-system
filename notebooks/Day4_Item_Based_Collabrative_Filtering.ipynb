```python
import pandas as pd
```

```python
rating_cols = [ 'user_id', 'movie_id', 'rating', 'timestamp' ]

ratings = pd.read_csv ( '../data/ml-100k/u.data', sep = '\t', names = rating_cols )

movie_cols = [ 'movie_id', 'title' ]

movies = pd.read_csv ( '../data/ml-100k/u.item', sep = '|', header = None, encoding = 'latin-1', usecols = [0, 1], names = movie_cols )

df = pd.merge ( ratings, movies, on = 'movie_id' )

user_movie_matrix = df.pivot_table ( index = 'user_id', columns = 'title', values = 'rating' )

user_movie_matrix.head()
```

```python
# Movie < - > Movie relationship

# Selecting one movie

batman_ratings = user_movie_matrix['Evil Dead II (1987)']

similar_movies = user_movie_matrix.corrwith ( batman_ratings )  # Find similar movies

# Clean results (Just correlation with every movies list)

corr_batman = pd.DataFrame ( similar_movies, columns = ['Correlation'] )

corr_batman.dropna ( inplace = True )

corr_batman.head()
```

```python
# Sort similar movies

corr_batman.sort_values ( 'Correlation', ascending = False ).head(10)
```

```python
# Add Rating counts

ratings_count = df.groupby ( 'title' )[ 'rating' ].count()

corr_batman  = corr_batman.join ( ratings_count )

corr_batman.columns = [ 'correlation', 'num_ratings' ]

corr_batman.head()
```

```python
# Filter weak movies

corr_batman [ corr_batman [ 'num_ratings' ] > 100 ]. sort_values ( 'correlation', ascending = False ).head()
```
