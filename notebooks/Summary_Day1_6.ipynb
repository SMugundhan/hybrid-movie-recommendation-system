🚀 Day 2 — Popularity-Based Recommendation

Created:

movie_stats = df.groupby(
    'title'
).agg(
    avg_rating=('rating','mean'),
    num_rating=('rating','count')
)

Problem:

Movies with few ratings may get artificially high scores.

Solution:

Weighted Rating:

WR=\frac{v}{v+m}R+\frac{m}{v+m}C

Where:

v → number of ratings

R → average movie rating

m → minimum threshold

C → overall average rating


Learned:

Higher thresholds increase reliability

Popularity ≠ personalization

🚀 Day 4 — Item-Based Collaborative Filtering

Shift:

Discovered issue:

Spurious correlation

Example:

Romance/Drama

Reason:

Very few overlapping ratings

Solution:

num_ratings>100

Key lesson:

Similarity alone is not enough.

Need:

Similarity + sufficient evidence
